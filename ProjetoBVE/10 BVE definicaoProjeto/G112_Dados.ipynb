## 1. Apresentação


Os dados abaixo parametrizam os três submodelos descritos em G112_Modelo_Conceitual (navios, barcaças e PCO) e permitem validar o cenário-base contra a operação real. Pedimos os registros brutos (planilhas ou extrações de sistema), não médias, dos últimos 12 a 24 meses. Quando não houver registro, pedimos a estimativa da equipe da BVE em três valores: mínimo, mais provável e máximo, que usaremos em distribuições triangulares (Law, 2013). Os itens críticos da Seção 6 devem chegar até o fim da semana 2.
