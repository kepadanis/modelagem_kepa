## 1. Formulação do problema

A Buena Vista Energy abastece navios no Porto Mar Azul com uma frota de seis barcaças de 3.000 t. As barcaças são carregadas na base da BVE, que tem dois berços, e seguem pelo canal até os navios em operação nos terminais do porto, onde o combustível é transferido por bombeamento. A programação das barcaças cabe à equipe de programação e controle da operação (PCO).

A demanda está crescendo, e a BVE precisa decidir como expandir em 20% a capacidade do sistema da forma técnica e economicamente mais vantajosa. As alternativas atuam sobre recursos que interagem: uma barcaça adicional pode apenas aumentar a fila na base se os berços já forem o gargalo, e novos berços pouco ajudam se as barcaças passam a maior parte do ciclo navegando ou bombeando.

O abastecimento só ocorre enquanto o navio está em operação no terminal, e não altera essa operação: o navio deixa o porto no horário previsto, abastecido ou não. Um pedido que a frota não consegue concluir dentro dessa janela é uma venda **perdida**. Em operação semelhante, a programação é refeita duas vezes ao dia (Costa e Mesquita, 2023).

Chegadas aleatórias, janelas de atendimento, tempos variáveis e recursos compartilhados tornam pouco confiáveis as estimativas por cálculo estático. A simulação de eventos discretos reproduz essa dinâmica, permite testar alternativas sem interferir na operação e compara os resultados com base estatística.

## 2. Objetivo e escopo

**Objetivo geral:** identificar, por simulação de eventos discretos, a configuração de menor custo que eleve em pelo menos 20% a capacidade do sistema de abastecimento da BVE, mantendo o nível de serviço de referência.

**Perguntas do estudo**

1. Qual é a capacidade atual do sistema e qual recurso a limita?
2. Quais alternativas, isoladas ou combinadas, atingem +20% de capacidade?
3. Qual delas tem o menor custo por tonelada de capacidade adicional?
4. Quanto se ganha apenas com mudanças nas regras do PCO, sem investimento em ativos?
5. Quão robusta é a recomendação a variações de demanda e à indisponibilidade de barcaças?

**Capacidade.** O nível de serviço é a fração dos pedidos atendidos dentro da janela de operação do navio; a referência será definida com a BVE ou, na falta dela, o valor do cenário-base. A demanda simulada é multiplicada por um fator *f_d* (1,0 a 1,4), mantendo o perfil de navios e volumes. A capacidade é o maior *f_d* em que o nível de serviço permanece na referência; uma alternativa cumpre a meta se atingir *f_d* ≥ 1,2.

## 3. Método

O estudo segue as etapas clássicas de projetos de simulação (Law, 2015; Chwif e Medina, 2015). O modelo será implementado em Python com a biblioteca SimPy, de forma incremental, em três submodelos ligados por arquivos CSV:

| Submodelo | Representa | Uso |
|:--|:--|:--|
| 1. Navios | Chegadas, fila no fundeio e operação nos terminais | Gera as janelas e os pedidos; validado contra as escalas do porto |
| 2. Barcaças | Ciclo carga - navegação - bombeamento - retorno, com demanda sorteada | Mede a capacidade da frota e da base isoladamente |
| 3. Integrado (PCO) | Programação das barcaças para os pedidos do submodelo 1 | Avalia nível de serviço, perdas e alternativas |

As distribuições de entrada serão ajustadas aos registros da BVE; na falta deles, usaremos estimativas de mínimo, mais provável e máximo em distribuições triangulares (Law, 2013). A validação compara o cenário-base com os registros históricos, com apoio dos programadores do PCO. Todo o material ficará versionado no GitHub e será entregue integralmente à BVE.

## 7. Premissas, restrições e riscos


**Premissas**, a confirmar com a BVE:

- o abastecimento ocorre durante a operação do navio no terminal e não a altera; pedido não concluído na janela é perdido;
- cada pedido é atendido por uma única barcaça;
- a operação é contínua, 24 horas por dia;
- o estoque da base não limita o carregamento;
- o crescimento mantém o perfil atual de navios, volumes e terminais;
- sem registros da BVE, tempos e vazões serão tomados de operação análoga (Costa e Mesquita, 2023).

**Restrições:** 10 semanas e 240 h de esforço; dados limitados ao que a BVE fornecer; custos de investimento dependentes de informação do cliente.

**Riscos**

| Risco | Impacto | Mitigação |
|:--|:--|:--|
| Escalas e pedidos registrados de forma incompleta | Demanda mal caracterizada e validação fraca | Solicitar amostra na semana 1; dados AIS; estimativas triangulares; sensibilidade |
| Regra do PCO manual e não documentada | Cenário-base não reproduz a operação | Entrevista com os programadores e comparação com programas emitidos |
| Custos de berços, barcaças e bombas indisponíveis | Comparação econômica inconclusiva | Faixas de custo de mercado e análise de sensibilidade |
| Indisponibilidade de barcaças sem histórico | Capacidade superestimada | Cenários com uma barcaça fora de operação |
